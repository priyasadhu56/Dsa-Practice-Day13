```python
# Move all positive numbers first
arr = [1, -2, 3, -4, 5, -6, 7, -8]
positive_numbers = [num for num in arr if num > 0]
negative_numbers = [num for num in arr if num < 0]
result = positive_numbers + negative_numbers
print(result) 
```

```python
# Move all negetive numbers first
arr = [1, -2, 3, -4, 5, -6, 7, -8]
negative_numbers = [num for num in arr if num < 0]
positive_numbers = [num for num in arr if num > 0]
result = negative_numbers + positive_numbers
print(result)
```

```python
# Left rotate by one position
arr = [1,7,9,4,3,2]
pos = arr[0]
for i in range(len(arr)-1):
    arr[i] = arr[i+1]
arr[len(arr)-1] = pos
print(arr)
```
